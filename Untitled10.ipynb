```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras import layers, models
from tensorflow.keras.datasets import fashion_mnist
```

```python
(x_train, _), (x_test, _) = fashion_mnist.load_data()

print(x_train.shape)
print(x_test.shape)
```

```python
plt.figure(figsize=(10, 3))

for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_train[i], cmap="gray")
    plt.axis("off")

plt.show()
```

```python
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

x_train = np.expand_dims(x_train, axis=-1)
x_test = np.expand_dims(x_test, axis=-1)

print(x_train.shape)
```

```python
noise_factor = 0.15

x_train_noisy = x_train + noise_factor * np.random.normal(
    loc=0.0,
    scale=1.0,
    size=x_train.shape
)

x_test_noisy = x_test + noise_factor * np.random.normal(
    loc=0.0,
    scale=1.0,
    size=x_test.shape
)

x_train_noisy = np.clip(x_train_noisy, 0.0, 1.0)
x_test_noisy = np.clip(x_test_noisy, 0.0, 1.0)
```

```python
plt.figure(figsize=(12, 4))

for i in range(5):

    plt.subplot(2, 5, i + 1)
    plt.imshow(x_test[i].squeeze(), cmap="gray")
    plt.axis("off")

    plt.subplot(2, 5, i + 6)
    plt.imshow(x_test_noisy[i].squeeze(), cmap="gray")
    plt.axis("off")

plt.show()
```

```python
autoencoder = models.Sequential([

    layers.Input(shape=(28, 28, 1)),

    # Encoder
    layers.Conv2D(32, (3, 3), activation="relu", padding="same"),
    layers.MaxPooling2D((2, 2), padding="same"),

    layers.Conv2D(64, (3, 3), activation="relu", padding="same"),
    layers.MaxPooling2D((2, 2), padding="same"),

    # Decoder
    layers.Conv2DTranspose(
        64,
        (3, 3),
        strides=2,
        activation="relu",
        padding="same"
    ),

    layers.Conv2DTranspose(
        32,
        (3, 3),
        strides=2,
        activation="relu",
        padding="same"
    ),

    layers.Conv2D(
        1,
        (3, 3),
        activation="sigmoid",
        padding="same"
    )
])
```

```python
autoencoder.summary()
```

```python
autoencoder.compile(
    optimizer="adam",
    loss="mse"
)
```

```python
history = autoencoder.fit(
    x_train_noisy,
    x_train,
    epochs=10,
    batch_size=128,
    validation_data=(x_test_noisy, x_test)
)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")

plt.xlabel("Epoch")
plt.ylabel("MSE Loss")
plt.legend()
plt.show()
```

```python
decoded_images = autoencoder.predict(x_test_noisy)
```

```python
n = 8

plt.figure(figsize=(16, 6))

for i in range(n):

    # Original
    plt.subplot(3, n, i + 1)
    plt.imshow(x_test[i].squeeze(), cmap="gray")
    plt.axis("off")

    # Noisy
    plt.subplot(3, n, i + 1 + n)
    plt.imshow(x_test_noisy[i].squeeze(), cmap="gray")
    plt.axis("off")

    # Denoised
    plt.subplot(3, n, i + 1 + 2*n)
    plt.imshow(decoded_images[i].squeeze(), cmap="gray")
    plt.axis("off")

plt.show()
```

```python
mse = np.mean((x_test - decoded_images) ** 2)

print("MSE:", mse)
```

```python
from skimage.metrics import structural_similarity as ssim

ssim_scores = []

for i in range(len(x_test)):
    score = ssim(
        x_test[i].squeeze(),
        decoded_images[i].squeeze(),
        data_range=1.0
    )
    ssim_scores.append(score)

average_ssim = np.mean(ssim_scores)

print("Average SSIM:", average_ssim)
```

```python
noise_levels = [0.05, 0.10, 0.15, 0.20, 0.30, 0.40]

results = []

for noise_level in noise_levels:

    # Add noise to test data
    noisy_test = x_test + noise_level * np.random.normal(
        loc=0.0,
        scale=1.0,
        size=x_test.shape
    )

    noisy_test = np.clip(noisy_test, 0.0, 1.0)

    # Denoise using the SAME trained model
    denoised = autoencoder.predict(
        noisy_test,
        verbose=0
    )

    # MSE
    mse_score = np.mean(
        (x_test - denoised) ** 2
    )

    # SSIM
    ssim_scores = []

    for i in range(len(x_test)):
        score = ssim(
            x_test[i].squeeze(),
            denoised[i].squeeze(),
            data_range=1.0
        )

        ssim_scores.append(score)

    average_ssim = np.mean(ssim_scores)

    results.append(
        [noise_level, mse_score, average_ssim]
    )

    print(
        f"Noise: {noise_level:.2f} | "
        f"MSE: {mse_score:.5f} | "
        f"SSIM: {average_ssim:.4f}"
    )
```

```python
import pandas as pd

results_df = pd.DataFrame(
    results,
    columns=["Noise Level", "MSE", "SSIM"]
)

results_df
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    results_df["Noise Level"],
    results_df["SSIM"],
    marker="o"
)

plt.xlabel("Noise Level")
plt.ylabel("Average SSIM")
plt.title("Denoising Performance vs Noise Level")
plt.grid()

plt.show()
```

```python
noise_levels = [0.05, 0.15, 0.20, 0.30, 0.40]

image_index = 0

plt.figure(figsize=(15, 6))

for i, noise_level in enumerate(noise_levels):

    # Take one clean image
    clean_image = x_test[image_index:image_index + 1]

    # Add noise
    noisy_image = clean_image + noise_level * np.random.normal(
        loc=0.0,
        scale=1.0,
        size=clean_image.shape
    )

    noisy_image = np.clip(noisy_image, 0.0, 1.0)

    # Let autoencoder remove the noise
    denoised_image = autoencoder.predict(
        noisy_image,
        verbose=0
    )

    # Show noisy image
    plt.subplot(2, len(noise_levels), i + 1)

    plt.imshow(
        noisy_image[0].squeeze(),
        cmap="gray"
    )

    plt.title(f"Noise {noise_level}")
    plt.axis("off")

    # Show denoised image
    plt.subplot(
        2,
        len(noise_levels),
        i + 1 + len(noise_levels)
    )

    plt.imshow(
        denoised_image[0].squeeze(),
        cmap="gray"
    )

    plt.title("Denoised")
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
small_autoencoder = models.Sequential([

    layers.Input(shape=(28, 28, 1)),

    # Encoder
    layers.Conv2D(
        16,
        (3, 3),
        activation="relu",
        padding="same"
    ),

    layers.MaxPooling2D(
        (2, 2),
        padding="same"
    ),

    layers.Conv2D(
        32,
        (3, 3),
        activation="relu",
        padding="same"
    ),

    layers.MaxPooling2D(
        (2, 2),
        padding="same"
    ),

    # Decoder
    layers.Conv2DTranspose(
        32,
        (3, 3),
        strides=2,
        activation="relu",
        padding="same"
    ),

    layers.Conv2DTranspose(
        16,
        (3, 3),
        strides=2,
        activation="relu",
        padding="same"
    ),

    layers.Conv2D(
        1,
        (3, 3),
        activation="sigmoid",
        padding="same"
    )
])

small_autoencoder.compile(
    optimizer="adam",
    loss="mse"
)

small_autoencoder.summary()
```

```python
print("Model A parameters:", autoencoder.count_params())
print("Model B parameters:", small_autoencoder.count_params())
```

```python
import time

start_time = time.time()

small_history = small_autoencoder.fit(
    x_train_noisy,
    x_train,
    epochs=10,
    batch_size=128,
    validation_data=(x_test_noisy, x_test)
)

small_training_time = time.time() - start_time

print("Training time:", small_training_time, "seconds")
```

```python
small_decoded = small_autoencoder.predict(
    x_test_noisy
)
```

```python
small_mse = np.mean(
    (x_test - small_decoded) ** 2
)

print("Small Model MSE:", small_mse)
```

```python
small_ssim_scores = []

for i in range(len(x_test)):

    score = ssim(
        x_test[i].squeeze(),
        small_decoded[i].squeeze(),
        data_range=1.0
    )

    small_ssim_scores.append(score)

small_ssim = np.mean(small_ssim_scores)

print("Small Model SSIM:", small_ssim)
```

```python
model_a_predictions = autoencoder.predict(
    x_test_noisy,
    verbose=0
)

model_a_mse = np.mean(
    (x_test - model_a_predictions) ** 2
)

model_a_ssim_scores = []

for i in range(len(x_test)):

    score = ssim(
        x_test[i].squeeze(),
        model_a_predictions[i].squeeze(),
        data_range=1.0
    )

    model_a_ssim_scores.append(score)

model_a_ssim = np.mean(model_a_ssim_scores)

print("Model A MSE:", model_a_mse)
print("Model A SSIM:", model_a_ssim)
```
